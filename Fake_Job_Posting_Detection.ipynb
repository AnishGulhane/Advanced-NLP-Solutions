```python
import string 
string.punctuation
```

```python
df['X'] = df['X'].str.lower()
```

```python
l1 = stopwords.words(('english'))
def text_process(mess):
    """
    1. remove the punctuation 
    2. remove the stopwords
    3. return the list of clean textwords
    """
    nopunc = [char for char in mess if char not in string.punctuation]
    nopunc = "".join(nopunc)

    return[ word for word in nopunc.split() if word not in l1 ]
```

```python
#df['X'].apply(text_process)# samjhane ke liye
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# assign a ID/token number to each word
```

```python



bow_transformer = CountVectorizer(analyzer=text_process).fit(df["X"])
```

```python
len(bow_transformer.vocabulary_)
```

```python
print(len(bow_transformer.vocabulary_))
```

```python
tdm = bow_transformer.transform(df['X'])
```

```python
tdm.shape
```

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(tdm, df.Y)

from sklearn.metrics import confusion_matrix, accuracy_score,recall_score,precision_score,f1_score
from sklearn.metrics import *

train_y = train_y.astype(int)
test_y = test_y.astype(int)
```

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression(class_weight = "balanced")
logreg.fit(train_x, train_y)#model build
```

```python
pred = logreg.predict(test_x)
print(confusion_matrix(test_y,pred))
print(classification_report(test_y, pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Limit max_depth and max_features to heavily speed up training
dt = DecisionTreeClassifier(max_depth=50, max_features='sqrt',class_weight = "balanced")
dt.fit(train_x, train_y)

pred = dt.predict(test_x)
tab = confusion_matrix(test_y, pred)
print(tab)

print(accuracy_score(test_y, pred), " ", 
      recall_score(test_y, pred, average='weighted'), " ", 
      precision_score(test_y, pred, average='weighted'))
```

```python

from sklearn.ensemble import RandomForestClassifier

# Added n_estimators, max_depth, max_features, and n_jobs=-1
rf = RandomForestClassifier(n_estimators=50, max_depth=50, max_features='sqrt', n_jobs=-1, oob_score=True,class_weight = "balanced")
rf.fit(train_x, train_y)

pred = rf.predict(test_x)
tab = confusion_matrix(test_y, pred)
print(tab)

# Remember to use average='weighted' here too!
print(accuracy_score(test_y, pred))
```
